# **Mengatasi Missing Value**

Memisahkan kolom yang memiliki missing value lebih dari 75% dan kurang dari 75%.

```python
less = missing_values[missing_values < 1500].index # Mengambil nama kolom yang memiliki missing value < 1500
over = missing_values[missing_values >= 1500].index # Mengambil nama kolom yang memiliki missing value >= 1500
```

Contoh mengisi nilai yang hilang dengan median untuk kolom numerik

```python
# Ambil kolom numerik dari data training
numeric_features = (
    train[less]                       # ambil kolom tertentu dari data training
    .select_dtypes(include=['number']) # pilih hanya kolom bertipe angka
    .columns                            # ambil nama kolomnya saja
)

# Mengisi nilai kosong (NaN) pada kolom numerik
# dengan nilai median dari masing-masing kolom
train[numeric_features] = (
    train[numeric_features]                    # ambil kolom numerik saja
    .fillna(train[numeric_features].median())  # isi NaN dengan median kolom
)
```

Contoh mengisi nilai yang hilang dengan mode untuk kolom kategori

```python
# Mengambil nama-nama kolom bertipe kategori (teks/object)
# dari dataset train pada kolom yang dipilih (less)
kategorical_features = (
    train[less]                          # ambil subset kolom
    .select_dtypes(include=['object'])   # pilih kolom bertipe teks/kategori
    .columns                             # ambil nama kolomnya saja
)

# Melakukan perulangan pada setiap kolom kategori
for column in kategorical_features:
    
    # Mengisi nilai kosong (NaN) pada kolom tersebut
    # dengan nilai yang paling sering muncul (modus)
    train[column] = train[column].fillna(
        train[column].mode()[0]
    )
```

```python
df = train.drop(columns=over) # Menghapus kolom dengan terlalu banyak nilai yang hilang
```

# **Melatih Model (Training)**

```python
print("Cek tipe data fitur pada x_train:")
print(x_train.dtypes ,"\n")
print("Cek tipe data target pada y_train:")
print(y_train.dtypes)
```

```python
# Melatih model 1 dengan algoritma Least Angle Regression
from sklearn import linear_model
lars = linear_model.Lars(n_nonzero_coefs=1).fit(x_train, y_train)

# Melatih model 2 dengan algoritma Linear Regression
from sklearn.linear_model import LinearRegression
LR = LinearRegression().fit(x_train, y_train)

# Melatih model 3 dengan algoritma Gradient Boosting Regressor
from sklearn.ensemble import GradientBoostingRegressor
GBR = GradientBoostingRegressor(random_state=184)
GBR.fit(x_train, y_train)
```

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
 
# Evaluasi pada model LARS
pred_lars = lars.predict(x_test)
mae_lars = mean_absolute_error(y_test, pred_lars)
mse_lars = mean_squared_error(y_test, pred_lars)
r2_lars = r2_score(y_test, pred_lars)
 
# Membuat dictionary untuk menyimpan hasil evaluasi
data = {
    'MAE': [mae_lars],
    'MSE': [mse_lars],
    'R2': [r2_lars]
}
 
# Konversi dictionary menjadi DataFrame
df_results = pd.DataFrame(data, index=['Lars'])
df_results
```

```python
# Evaluasi pada model Linear Regression
pred_LR = LR.predict(x_test)
mae_LR = mean_absolute_error(y_test, pred_LR)
mse_LR = mean_squared_error(y_test, pred_LR)
r2_LR = r2_score(y_test, pred_LR)
 
# Menambahkan hasil evaluasi LR ke DataFrame
df_results.loc['Linear Regression'] = [mae_LR, mse_LR, r2_LR]
df_results
```

```python
# Evaluasi pada model Gradient Boosting Regressor
pred_GBR = GBR.predict(x_test)
mae_GBR = mean_absolute_error(y_test, pred_GBR)
mse_GBR = mean_squared_error(y_test, pred_GBR)
r2_GBR = r2_score(y_test, pred_GBR)
 
# Menambahkan hasil evaluasi GBR ke DataFrame
df_results.loc['GradientBoostingRegressor'] = [mae_GBR, mse_GBR, r2_GBR]
df_results
```
